```python
from datasets import load_dataset

url_gpt4 = "https://github.com/paul-rottger/exaggerated-safety/raw/main/model_completions/xstest_v2_completions_gpt4.csv"
url_llama2new = "https://github.com/paul-rottger/exaggerated-safety/raw/main/model_completions/xstest_v2_completions_llama2new.csv"
url_llama2orig = "https://github.com/paul-rottger/exaggerated-safety/raw/main/model_completions/xstest_v2_completions_llama2orig.csv"
url_mistralguard = "https://github.com/paul-rottger/exaggerated-safety/raw/main/model_completions/xstest_v2_completions_mistralguard.csv"
url_mistralinstruct = "https://github.com/paul-rottger/exaggerated-safety/raw/main/model_completions/xstest_v2_completions_mistralinstruct.csv"

url_prompts = "https://github.com/paul-rottger/exaggerated-safety/raw/main/xstest_v2_prompts.csv"
```

```python
# for each url, load with huggingface load_dataset from csv
data_gpt4 = load_dataset("csv", data_files=url_gpt4)["train"]
data_llama2new = load_dataset("csv", data_files=url_llama2new)["train"]
data_llama2orig = load_dataset("csv", data_files=url_llama2orig)["train"]
data_mistralguard = load_dataset("csv", data_files=url_mistralguard)["train"]
data_mistralinstruct = load_dataset("csv", data_files=url_mistralinstruct)["train"]

data_prompts = load_dataset("csv", data_files=url_prompts)["train"]
```

```python
# move columns in the prompts so they have the same structure as the other datasets
# remove id_v1, rename id_v2 as id, empty column completion, focus to annotation_1 and note to annotation_2
# data_prompts_backup = data_prompts # for notebooking and not reloading
# data_prompts = data_prompts_backup
data_prompts = data_prompts.rename_column("id_v2", "id")
data_prompts = data_prompts.rename_column("focus", "annotation_1")
data_prompts = data_prompts.rename_column("note", "annotation_2")
# remove id_v1
data_prompts = data_prompts.remove_columns("id_v1")

# add empty columns for agreement, completion, and final_label
data_prompts = data_prompts.add_column("agreement", [False]*len(data_prompts))
data_prompts = data_prompts.add_column("final_label", [""]*len(data_prompts))
data_prompts = data_prompts.add_column("completion", [""]*len(data_prompts))

# Create a dataset dict with each subset of the data from above
from datasets import DatasetDict
dataset = DatasetDict({
    "gpt4": data_gpt4,
    "llama2new": data_llama2new,
    "llama2orig": data_llama2orig,
    "mistralguard": data_mistralguard,
    "mistralinstruct": data_mistralinstruct,
    "prompts": data_prompts
})
```
